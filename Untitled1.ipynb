```python
import cv2
import matplotlib.pyplot as plt

# Image Path
image_path = '/content/WhatsApp Image 2026-11-05 at 11.19.43 AM.jpeg'

# Load Colour Image
color_img = cv2.imread(image_path)

# Load Grayscale Image
gray_img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

# Check Image Information
print("Colour Image Shape:", color_img.shape)
print("Colour Image Data Type:", color_img.dtype)

print("Grayscale Image Shape:", gray_img.shape)
print("Grayscale Image Data Type:", gray_img.dtype)
```

```python
# Create Display Function

def show(img, title):

    plt.figure(figsize=(7, 5))

    if len(img.shape) == 3:
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        plt.imshow(img)
    else:
        plt.imshow(img, cmap='gray')

    plt.title(title)
    plt.axis('off')
    plt.show()


# Display Images
show(color_img, "Original Colour Image")
show(gray_img, "Grayscale Image")
```

```python
# Convert Grayscale Image into Binary Image

ret, binary_img = cv2.threshold(
    gray_img,
    127,
    255,
    cv2.THRESH_BINARY
)

print("Threshold Value:", ret)
print("Binary Image Shape:", binary_img.shape)
print("Binary Image Data Type:", binary_img.dtype)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(gray_img, cmap='gray')
plt.title("Original Grayscale Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(binary_img, cmap='gray')
plt.title("Binary Image")
plt.axis('off')

plt.show()
```

```python
# Create 5x5 Rectangular Kernel

kernel = np.ones((5, 5), np.uint8)

print("Kernel:")
print(kernel)
```

```python
# Erosion
erosion = cv2.erode(binary_img, kernel, iterations=1)

# Dilation
dilation = cv2.dilate(binary_img, kernel, iterations=1)

# Opening
opening = cv2.morphologyEx(
    binary_img,
    cv2.MORPH_OPEN,
    kernel
)

# Closing
closing = cv2.morphologyEx(
    binary_img,
    cv2.MORPH_CLOSE,
    kernel
)

print("All Four Morphological Operations Completed!")
```

```python
plt.figure(figsize=(18, 5))

images = [
    binary_img,
    erosion,
    dilation,
    opening,
    closing
]

titles = [
    "Original Binary",
    "Erosion",
    "Dilation",
    "Opening",
    "Closing"
]

for i in range(5):
    plt.subplot(1, 5, i + 1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
# Define Kernel Shapes

kernel_shapes = {
    'RECT': cv2.MORPH_RECT,
    'ELLIPSE': cv2.MORPH_ELLIPSE,
    'CROSS': cv2.MORPH_CROSS
}

kernel_sizes = [3, 5, 9]
```

```python
# Create 3x3 Comparison Grid

plt.figure(figsize=(12, 12))

plot_index = 1

for shape_name, shape in kernel_shapes.items():

    for size in kernel_sizes:

        kernel = cv2.getStructuringElement(
            shape,
            (size, size)
        )

        # Apply Opening
        result = cv2.morphologyEx(
            binary_img,
            cv2.MORPH_OPEN,
            kernel
        )

        plt.subplot(3, 3, plot_index)
        plt.imshow(result, cmap='gray')

        plt.title(f'{shape_name} - {size}x{size}')
        plt.axis('off')

        plot_index += 1

plt.tight_layout()
plt.show()
```

## Task 1.5 – Why Morphology Works

### 1. Erosion

Erosion is a local minimum filter. It shrinks the white regions of an image and removes small bright noise.

**Real-life Application:** Used to remove small unwanted white spots from scanned documents before text detection.

### 2. Dilation

Dilation is a local maximum filter. It expands white regions and fills small dark gaps in an image.

**Real-life Application:** Used to connect broken parts of characters in document processing.

### 3. Opening

Opening is performed by applying erosion followed by dilation. It removes small noise while preserving the overall shape and size of objects.

**Real-life Application:** Used to remove small unwanted particles from an image before object detection.

### 4. Closing

Closing is performed by applying dilation followed by erosion. It fills small holes and gaps without significantly changing the object's overall shape.

**Real-life Application:** Used to fill small gaps in detected objects before passing the mask to a downstream detector.

### Conclusion

Morphological operations improve image quality by removing unwanted noise, filling gaps and refining object boundaries. They help prepare images and masks for accurate downstream computer vision tasks.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

# YuNet model path
model_path = '/content/data/models/face_detection_yunet_2023mar.onnx'

# Create output folder
os.makedirs('/content/face_results', exist_ok=True)

# Process all uploaded PNG images
for image_name in uploaded.keys():

    # Read image
    img = cv2.imread(image_name)

    if img is None:
        print("Image not loaded:", image_name)
        continue

    height, width = img.shape[:2]

    # Create YuNet detector
    detector = cv2.FaceDetectorYN_create(
        model_path,
        "",
        (width, height),
        0.5,
        0.3,
        5000
    )

    detector.setInputSize((width, height))

    # Detect faces
    _, faces = detector.detect(img)

    result_img = img.copy()

    if faces is not None:
        print(image_name, "Faces Detected:", len(faces))

        for face in faces:

            # Bounding box
            x, y, w, h = face[:4].astype(int)

            cv2.rectangle(
                result_img,
                (x, y),
                (x + w, y + h),
                (0, 255, 0),
                2
            )

            # Confidence score
            confidence = face[-1]

            cv2.putText(
                result_img,
                f"Confidence: {confidence:.2f}",
                (x, max(y - 10, 20)),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.6,
                (0, 255, 0),
                2
            )

            # Five facial landmarks
            landmarks = face[4:14].reshape(5, 2).astype(int)

            for point in landmarks:
                cv2.circle(
                    result_img,
                    tuple(point),
                    4,
                    (0, 0, 255),
                    -1
                )

    else:
        print(image_name, "No Face Detected")

    # Display result
    plt.figure(figsize=(8, 6))
    plt.imshow(cv2.cvtColor(result_img, cv2.COLOR_BGR2RGB))
    plt.title("YuNet Detection - " + image_name)
    plt.axis("off")
    plt.show()

    # Save result
    output_path = '/content/face_results/' + image_name
    cv2.imwrite(output_path, result_img)

print("All images processed successfully!")
```

```python
# Face Blurring using Gaussian Blur

for image_name in uploaded.keys():

    img = cv2.imread(image_name)
    height, width = img.shape[:2]

    detector = cv2.FaceDetectorYN_create(
        model_path,
        "",
        (width, height),
        0.5,
        0.3,
        5000
    )

    detector.setInputSize((width, height))

    _, faces = detector.detect(img)

    blurred_img = img.copy()

    if faces is not None:

        print(image_name, "Faces Found:", len(faces))

        for face in faces:

            x, y, w, h = face[:4].astype(int)

            # Keep coordinates inside image boundaries
            x1 = max(0, x)
            y1 = max(0, y)
            x2 = min(width, x + w)
            y2 = min(height, y + h)

            # Extract face region
            face_region = blurred_img[y1:y2, x1:x2]

            # Apply Gaussian Blur
            blurred_face = cv2.GaussianBlur(
                face_region,
                (51, 51),
                0
            )

            blurred_img[y1:y2, x1:x2] = blurred_face

    else:
        print(image_name, "No face detected, cannot blur.")

    # Display original and blurred image
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(blurred_img, cv2.COLOR_BGR2RGB))
    plt.title("Face Blurred Image")
    plt.axis("off")

    plt.show()

    # Save blurred image
    cv2.imwrite(
        '/content/face_results/blurred_' + image_name,
        blurred_img
    )
```

```python
import cv2
import time
from IPython.display import display, Video

video_path = '/content/yolo_test_video.mp4'
output_path = '/content/yolo_detected_video.mp4'

# Open video
cap = cv2.VideoCapture(video_path)

# Video properties
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps_original = cap.get(cv2.CAP_PROP_FPS)

# Create video writer
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_path, fourcc, fps_original, (width, height))

frame_count = 0
start_time = time.time()

while cap.isOpened():

    ret, frame = cap.read()

    if not ret:
        break

    # YOLO detection
    results = model.predict(frame, conf=0.25, verbose=False)

    # Draw bounding boxes
    detected_frame = results[0].plot()

    out.write(detected_frame)

    frame_count += 1

cap.release()
out.release()

end_time = time.time()

# Calculate FPS
total_time = end_time - start_time
average_fps = frame_count / total_time

print("Total Frames:", frame_count)
print("Processing Time:", round(total_time, 2), "seconds")
print("Average FPS:", round(average_fps, 2))

print("Video Processing Completed!")
```

```python
import cv2
import time
import os

# Input and output video
video_path = '/content/yolo_test_video.mp4'
output_path = '/content/integrated_pipeline.mp4'

# Open video
cap = cv2.VideoCapture(video_path)

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)

# Create YuNet detector
face_detector = cv2.FaceDetectorYN_create(
    model_path,
    "",
    (width, height),
    0.5,
    0.3,
    5000
)

face_detector.setInputSize((width, height))

# Create output video
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(
    output_path,
    fourcc,
    fps,
    (width, height)
)

frame_count = 0
start_time = time.time()

while cap.isOpened():

    ret, frame = cap.read()

    if not ret or frame_count >= 100:
        break

    # 1. YuNet Face Detection
    _, faces = face_detector.detect(frame)

    if faces is not None:
        for face in faces:
            x, y, w, h = face[:4].astype(int)

            cv2.rectangle(
                frame,
                (x, y),
                (x + w, y + h),
                (0, 255, 0),
                2
            )

            cv2.putText(
                frame,
                "FACE",
                (x, max(y - 10, 20)),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.6,
                (0, 255, 0),
                2
            )

    # 2. YOLO Object Detection
    results = model.predict(
        frame,
        conf=0.25,
        verbose=False
    )

    for box in results[0].boxes:

        x1, y1, x2, y2 = box.xyxy[0].int().tolist()

        class_id = int(box.cls[0])
        label = model.names[class_id]

        cv2.rectangle(
            frame,
            (x1, y1),
            (x2, y2),
            (0, 0, 255),
            2
        )

        cv2.putText(
            frame,
            label,
            (x1, max(y1 - 10, 20)),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (0, 0, 255),
            2
        )

    out.write(frame)
    frame_count += 1

cap.release()
out.release()

total_time = time.time() - start_time

print("Integrated Pipeline Completed!")
print("Total Frames Processed:", frame_count)
print("Processing Time:", round(total_time, 2), "seconds")
print("Output Saved:", output_path)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read one frame from video
cap = cv2.VideoCapture('/content/yolo_test_video.mp4')
ret, original_frame = cap.read()
cap.release()

if not ret:
    print("Video frame could not be read.")
else:
    # Add light salt-and-pepper noise
    noisy_frame = original_frame.copy()

    noise = np.random.random(noisy_frame.shape[:2])

    noisy_frame[noise < 0.01] = 0
    noisy_frame[noise > 0.99] = 255

    # Apply morphological opening
    kernel = np.ones((3, 3), np.uint8)

    cleaned_frame = cv2.morphologyEx(
        noisy_frame,
        cv2.MORPH_OPEN,
        kernel
    )

    # Function to count detections
    def count_detections(frame):

        h, w = frame.shape[:2]

        face_detector.setInputSize((w, h))
        _, faces = face_detector.detect(frame)

        face_count = 0 if faces is None else len(faces)

        results = model.predict(
            frame,
            conf=0.25,
            verbose=False
        )

        object_count = len(results[0].boxes)

        return face_count, object_count

    # Compare both frames
    noisy_faces, noisy_objects = count_detections(noisy_frame)
    clean_faces, clean_objects = count_detections(cleaned_frame)

    print("WITHOUT Morphological Opening")
    print("Faces:", noisy_faces)
    print("Objects:", noisy_objects)

    print("\nWITH Morphological Opening")
    print("Faces:", clean_faces)
    print("Objects:", clean_objects)

    # Display comparison
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(noisy_frame, cv2.COLOR_BGR2RGB))
    plt.title("Noisy Frame")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(cleaned_frame, cv2.COLOR_BGR2RGB))
    plt.title("After Morphological Opening")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

# Dynamic self-healing: Find the image file, or fallback to generating a clean binary shape canvas
if 'binary_img' not in globals():
    print("Recreating 'binary_img'...")

    # Scan workspace for the uploaded file
    possible_files = [f for f in os.listdir('/content') if 'WhatsApp Image' in f and f.endswith(('.jpeg', '.jpg', '.png'))]

    if possible_files:
        image_path = os.path.join('/content', possible_files[-1])
        print(f"Found source image: {image_path}")
        gray_img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
        ret, binary_img = cv2.threshold(gray_img, 127, 255, cv2.THRESH_BINARY)
    else:
        print("No matching WhatsApp image found in /content. Generating an elegant fallback canvas...")
        # Create a synthetic high-contrast binary image (white shapes on a black background)
        fallback = np.zeros((400, 400), dtype=np.uint8)
        cv2.circle(fallback, (150, 200), 70, 255, -1)
        cv2.rectangle(fallback, (220, 130), (330, 270), 255, -1)
        cv2.putText(fallback, "CV", (50, 350), cv2.FONT_HERSHEY_SIMPLEX, 4, 255, 10)
        binary_img = fallback

# Create a noisy binary image
noisy_binary = binary_img.copy()

# Add salt and pepper noise
noise = np.random.random(noisy_binary.shape)

noisy_binary[noise < 0.02] = 255
noisy_binary[noise > 0.98] = 0

# Define kernel shapes and sizes
kernel_shapes = {
    "RECT": cv2.MORPH_RECT,
    "ELLIPSE": cv2.MORPH_ELLIPSE,
    "CROSS": cv2.MORPH_CROSS
}

kernel_sizes = [3, 5, 9]

# Display noisy image
plt.figure(figsize=(5, 5))
plt.imshow(noisy_binary, cmap='gray')
plt.title("Noisy Binary Image")
plt.axis("off")
plt.show()

# Compare Opening
plt.figure(figsize=(12, 10))

plot_number = 1

for shape_name, shape in kernel_shapes.items():
    for size in kernel_sizes:

        kernel = cv2.getStructuringElement(
            shape, (size, size)
        )

        result = cv2.morphologyEx(
            noisy_binary, cv2.MORPH_OPEN, kernel
        )

        plt.subplot(3, 3, plot_number)
        plt.imshow(result, cmap='gray')
        plt.title(f"{shape_name} - {size}x{size}")
        plt.axis("off")

        plot_number += 1

plt.suptitle("Morphological Opening Comparison")
plt.tight_layout()
plt.savefig("/content/plots/morphology_opening_grid.png")
plt.show()


# Compare Closing
plt.figure(figsize=(12, 10))

plot_number = 1

for shape_name, shape in kernel_shapes.items():
    for size in kernel_sizes:

        kernel = cv2.getStructuringElement(
            shape, (size, size)
        )

        result = cv2.morphologyEx(
            noisy_binary, cv2.MORPH_CLOSE, kernel
        )

        plt.subplot(3, 3, plot_number)
        plt.imshow(result, cmap='gray')
        plt.title(f"{shape_name} - {size}x{size}")
        plt.axis("off")

        plot_number += 1

plt.suptitle("Morphological Closing Comparison")
plt.tight_layout()
plt.savefig("/content/plots/morphology_closing_grid.png")
plt.show()

print("Opening and Closing comparisons completed!")
```
